# **Linear Regression**

## 1. Introduction

Linear Regression is a supervised machine learning algorithm
used to model the relationship between an independent variable
(input) and a dependent variable (output).

In this implementation, we explore the relationship between
MedHouseVal and MedInc using the California Housing Dataset.

```python
# importing important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression
```

## 2. Dataset Description

### Dataset: California Housing Dataset

The California Housing Dataset contains information about
housing districts in California.

It includes demographic and geographical information,
along with housing-related measurements.

**Dataset Details:**
- Number of samples: 20,640
- Number of features: 8
- Target column included in the dataset: MedHouseVal
- Total columns in the DataFrame: 9



## 3. Data Preparation

The dataset is loaded using the `fetch_california_housing()`
function from Scikit-learn.

The `as_frame=True` parameter loads the dataset in
Pandas DataFrame format.

The `.frame` attribute provides the complete dataset,
including its features and target column.

The `head()` function displays the first five rows.

```python
# loading the data
california_data = fetch_california_housing(as_frame=True)

df = california_data.frame
df.head()

```

```python
# Display the dataset feature names
df.columns
```

## 4. Feature Selection

For this implementation, the following variables are selected:

- Independent variable (X): MedHouseVal
- Dependent variable (y): MedInc

The objective is to use MedHouseVal as the input
to predict MedInc.

Note: This variable selection follows the assignment instructions.

```python
# Select the input and target variables
x=df["MedHouseVal"]
y=df["MedInc"]
print(x.shape)
print(y.shape)
```

## 5. Data Visualization

A scatter plot is used to visualize the relationship
between the independent and dependent variables.

Each point represents one observation from the dataset.

The x-axis represents MedHouseVal, while the
y-axis represents MedInc.

```python
plt.scatter(x,y)
plt.xlabel("MedHouseVal")
plt.ylabel("MedInc")
plt.show()
```

## 6. Model Training

A Linear Regression model is created using Scikit-learn.

The input variable is reshaped into a two-dimensional
array because Scikit-learn expects input features in
the format (number of samples, number of features).

The `fit()` method trains the model using the input
and target variables.

The slope and intercept are retrieved using:
- `coef_`: Model coefficient (slope)
- `intercept_`: Model intercept

```python
X=x.values.reshape(-1,1)
model=LinearRegression()
model.fit(X,y)
print("Slope:",model.coef_)
print("Intercept:",model.intercept_)
```

## 7. Regression Line Visualization

The regression line is plotted over the scatter plot to visualize the predictions made by the model.

```python
# Generate predictions
y_pred=model.predict(X)

# np.asarray() converts the values into a NumPy array.
# ravel() converts the array into a one-dimensional array.
x_plot=np.asarray(X).ravel()
y_pred_plot=np.asarray(y_pred).ravel()

# np.argsort() returns the indices that arrange the input values in ascending order.
indices=np.argsort(x_plot)

# Plot the data and regression line
plt.scatter(x_plot,y)
plt.plot(x_plot[indices],y_pred_plot[indices],color='orange')
plt.xlabel("MedHouseVal")
plt.ylabel("MedInc")
plt.show()
```

## 8. Results and Observations

- The model shows a positive relationship between MedHouseVal and MedInc.
- The slope is approximately 1.1328, indicating that the predicted MedInc increases as MedHouseVal increases.
- The intercept is approximately 1.5274, representing the predicted MedInc when MedHouseVal is zero.
- The regression line captures the general trend, but the actual data points are widely scattered around it.
